```python
from datasets import load_from_disk
import os
```

```python
# Caminho para os datasets processados localmente
# A estrutura do cookiecutter sugere 'data/processed'
processed_data_path = '../data/processed/'

# Carrega os subconjuntos locais
bookcorpus_local = load_from_disk(os.path.join(processed_data_path, 'bookcorpus_subset'))
wiki_local = load_from_disk(os.path.join(processed_data_path, 'wiki_subset'))

print("Subconjuntos locais carregados com sucesso!")
print(bookcorpus_local)
print(wiki_local)

# Agora você pode trabalhar com eles normalmente
print("\n--- Exemplo de dado do BookCorpus Local ---")
print(bookcorpus_local[0])
```
