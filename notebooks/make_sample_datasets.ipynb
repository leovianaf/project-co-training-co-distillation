# Retirar uma amostra dos Datasets para testes

Os datasets `BookCorpus`e `Wikipedia` originais são muito grandes, então foi necessário rodar o notebook abaixo no **Google Colab** para conseguir uma amostra de demonstração para montagem do pipeline completo.

```python
from google.colab import drive

drive.mount('/content/drive')
```

```python
# Passo 1: Instalar a biblioteca necessária no ambiente do Colab
!pip install datasets

# Passo 2: Importar a biblioteca
from datasets import load_dataset
import os

# --- CARREGAR OS DATASETS COMPLETOS ---
print("Baixando BookCorpus...")
bookcorpus_full = load_dataset("rojagtap/bookcorpus", split='train')

print("\nBaixando Wikipedia...")
wiki_full = load_dataset("wikimedia/wikipedia", "20231101.en", split='train')

print("\nDatasets completos carregados!")
print(f"BookCorpus original com {len(bookcorpus_full)} exemplos.")
print(f"Wikipedia original com {len(wiki_full)} exemplos.")

# --- CRIAR SUBCONJUNTOS MENORES ---
subset_size_bookcorpus = 200  # 200 exemplos
subset_size_wiki = 100      # 100 exemplos

print(f"\nCriando subconjunto de {subset_size_bookcorpus} exemplos do BookCorpus...")
bookcorpus_subset = bookcorpus_full.shuffle(seed=42).select(range(subset_size_bookcorpus))

print(f"Criando subconjunto de {subset_size_wiki} exemplos da Wikipedia...")
wiki_subset = wiki_full.shuffle(seed=42).select(range(subset_size_wiki))

print("\nSubconjuntos criados com sucesso!")
print(bookcorpus_subset)
print(wiki_subset)

# --- SALVAR OS SUBCONJUNTOS NO DISCO DO COLAB ---
# A biblioteca 'datasets' salva em um formato otimizado (Arrow)
output_dir_bookcorpus = 'bookcorpus_subset'
output_dir_wiki = 'wiki_subset'

print(f"\nSalvando subconjunto do BookCorpus em '{output_dir_bookcorpus}'...")
bookcorpus_subset.save_to_disk(output_dir_bookcorpus)

print(f"Salvando subconjunto da Wikipedia em '{output_dir_wiki}'...")
wiki_subset.save_to_disk(output_dir_wiki)

print("\nTodos os subconjuntos foram salvos no ambiente do Colab!")
```

```python
# Compactar as pastas para facilitar o download
print("Compactando as pastas dos datasets...")
!zip -r bookcorpus_subset.zip bookcorpus_subset
!zip -r wiki_subset.zip wiki_subset
print("Compactação concluída!")

# Copiar os arquivos .zip para o seu Google Drive
# Eles serão salvos na pasta principal "Meu Drive".
print("\nCopiando arquivos para o Google Drive...")
!cp bookcorpus_subset.zip "/content/drive/MyDrive/CTCD"
!cp wiki_subset.zip "/content/drive/MyDrive/CTCD"
print("Arquivos copiados para o seu Google Drive com sucesso!")
```
